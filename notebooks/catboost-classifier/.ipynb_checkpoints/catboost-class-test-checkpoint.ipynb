```python
import gradio as gr
import pandas as pd
import torch
import torch.nn as nn
import json
import os
from catboost import CatBoostClassifier

# --- 1. АРХИТЕКТУРА ---
class TransformerRec(nn.Module):
    def __init__(self, vocab_size, hidden=256, heads=8, layers=4, max_len=20, dropout=0.2):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, hidden)
        self.pos_emb = nn.Embedding(max_len, hidden)
        self.dropout = nn.Dropout(dropout)
        encoder_layer = nn.TransformerEncoderLayer(d_model=hidden, nhead=heads, dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=layers)
        self.fc = nn.Linear(hidden, vocab_size)

    def forward(self, x):
        positions = torch.arange(x.size(1), device=x.device).unsqueeze(0)
        x = self.dropout(self.token_emb(x) + self.pos_emb(positions))
        x = self.transformer(x)
        return self.fc(x)

# --- 2. ЗАГРУЗКА (ОБНОВЛЕННАЯ) ---
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
data_folder = "../gym_model"
LEVEL_MAP = {'Novice': 0, 'Beginner': 1, 'Intermediate': 2, 'Advanced': 3}

with open(f'{data_folder}/vocab.json', 'r') as f:
    vocab = json.load(f)
id2name = {v: k for k, v in vocab.items()}

# Подтягиваем метаданные для проверки групп мышц
df_meta = pd.read_csv("../programs_detailed_boostcamp_kaggle.csv")
df_meta_unique = df_meta.drop_duplicates(subset=['exercise_name'])
exercise_meta_dict = df_meta_unique.set_index('exercise_name').to_dict('index')

# НОВОЕ: Подтягиваем ИНВЕНТАРЬ КАНДИДАТОВ из золотого словаря
df_equip = pd.read_csv("id_to_equipment_mapping_FIXED.csv") # Убедись, что путь правильный
id_to_eq_dict = df_equip.set_index('candidate_id')['equipment_golden'].to_dict()

model = TransformerRec(vocab_size=len(vocab)).to(device)
bert_path = f'{data_folder}/checkpoints-2/gym_bert_v2_ep27_hit0.3522.pth'
if os.path.exists(bert_path):
    model.load_state_dict(torch.load(bert_path, map_location=device))
    model.eval()

cat_ranker = CatBoostClassifier()

# Проверяем сначала текущую папку, затем data_folder
if os.path.exists("catboost_recommender_v1.cbm"):
    cat_path = "catboost_recommender_v1.cbm"
elif os.path.exists(f'{data_folder}/catboost_recommender_v1.cbm'):
    cat_path = f'{data_folder}/catboost_recommender_v1.cbm'
else:
    raise FileNotFoundError("🚨 ФАЙЛ НЕ НАЙДЕН: Убедись, что 'catboost_recommender_v1.cbm' лежит рядом со скриптом!")

cat_ranker.load_model(cat_path)
print(f"✅ Модель CatBoost успешно загружена из: {cat_path}")


# --- 3. ЧИСТАЯ ЛОГИКА ПРЕДСКАЗАНИЯ (ОБНОВЛЕННАЯ) ---
def predict_debug(history_ids, user_profile, top_k=10):
    input_seq = torch.LongTensor([history_ids]).to(device)
    with torch.no_grad():
        logits = model(input_seq)
        probs = torch.softmax(logits[0, -1, :], dim=-1)
    
    # Берем топ-100 от Трансформера
    scores, indices = torch.topk(probs, k=100) 
    candidates = []
    
    last_ex_name = id2name.get(history_ids[-1], "")
    last_meta = exercise_meta_dict.get(last_ex_name, {})

    for s, idx in zip(scores.cpu().numpy(), indices.cpu().numpy()):
        idx_val = int(idx)
        name = id2name[idx_val]
        current_meta = exercise_meta_dict.get(name, {})
        is_same = 1 if current_meta.get('category') == last_meta.get('category') else 0
        
        # ДОСТАЕМ РЕАЛЬНОЕ ОБОРУДОВАНИЕ КАНДИДАТА
        cand_eq = id_to_eq_dict.get(idx_val, 'Unknown')
        
        candidates.append({
            'candidate_id': str(idx_val),  # CatBoost ждет строку!
            'bert_score': float(s),
            'is_same_group': is_same,
            'level_idx': LEVEL_MAP.get(user_profile['level'], 1),
            'goal_clean': str(user_profile['goal']),
            'eq_clean': str(cand_eq)       # Имя колонки как в обучении!
        })
    
    cand_df = pd.DataFrame(candidates)
    
    # Строгий порядок колонок, как они были в X_train
    # ТОЧНЫЙ порядок колонок, как при обучении (X_train)
    features_order = ['bert_score', 'candidate_id', 'is_same_group', 'eq_clean', 'level_idx', 'goal_clean']
    
    # Предсказание CatBoost
    preds = cat_ranker.predict_proba(cand_df[features_order])[:, 1]
    cand_df['final_score'] = preds
    cand_df['name'] = [id2name[int(i)] for i in cand_df['candidate_id']]
    
    # --- ХАРД-ФИЛЬТР ПО ИНВЕНТАРЮ ЮЗЕРА ---
    user_eq = user_profile['equipment']
    if user_eq != 'Gym_Mixed': # Если не полный зал, жестко режем то, что не подходит
        # Разрешаем юзеру делать упражнения с выбранным инвентарем И с собственным весом
        cand_df = cand_df[cand_df['eq_clean'].isin([user_eq, 'Bodyweight', 'Cardio'])]
    
    # Сортируем и берем Топ-K
    final_df = cand_df.sort_values('final_score', ascending=False).head(top_k)
    
    cat_table = final_df[['name', 'final_score', 'bert_score', 'eq_clean']].copy()
    cat_table.columns = ['Упражнение', 'CatBoost Prob', 'BERT Score', 'Инвентарь']
    
    bert_df = pd.DataFrame({
        "Упражнение": [id2name[i.item()] for i in indices[:10]],
        "BERT Score": [round(s.item(), 4) for s in scores[:10]]
    })
    
    return final_df['name'].tolist(), bert_df, cat_table

# --- 4. ИНТЕРФЕЙС ---
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    history_state = gr.State([])
    profile_state = gr.State({})
    
    gr.Markdown("# 🏋️‍♂️ AI Trainer: Pure Model Test")
    
    with gr.Row(variant='panel'):
        level = gr.Dropdown(['Novice', 'Beginner', 'Intermediate', 'Advanced'], label="Уровень", value="Novice")
        goal = gr.Dropdown(['Bodybuilding', 'Muscle&Sculpting', 'Powerlifting', 'OlympicWeightlifting', 'BodyweightFitness'], label="Цель", value="Bodybuilding")
        equip = gr.Dropdown(['Bodyweight', 'Dumbbell', 'Barbell', 'Machine', 'Gym_Mixed'], label="Инвентарь", value="Bodyweight")
        start_btn = gr.Button("🚀 Настроить профиль", variant="primary")

    with gr.Column(visible=False) as main_app:
        history_display = gr.Textbox(label="Цепочка тренировки", interactive=False)
        
        with gr.Row() as first_ex_row:
            first_ex_drop = gr.Dropdown(choices=sorted(list(vocab.keys())), label="Выберите первое упражнение:", interactive=True)
            
        with gr.Row(visible=False) as results:
            with gr.Column():
                gr.Markdown("#### 🧠 BERT")
                bert_ui_table = gr.DataFrame()
            with gr.Column():
                gr.Markdown("#### ⚖️ CatBoost")
                cat_ui_table = gr.DataFrame()
        
        recommendations = gr.Dropdown(choices=[], label="Что дальше?", interactive=True, visible=False)
        add_btn = gr.Button("➕ Добавить", variant="primary", visible=False)
        reset_btn = gr.Button("🔄 Сброс", variant="stop")

    def start_fn(l, g, e):
        return {profile_state: {'level': l, 'goal': g, 'equipment': e}, main_app: gr.update(visible=True)}

    def add_ex_fn(ex_name, history, profile):
        if not ex_name: return history, "", gr.update(), None, None, gr.update(), gr.update()
        new_id = vocab[ex_name]
        if not history or history[-1] != new_id: history.append(new_id)
        
        recs, b_df, c_df = predict_debug(history, profile)
        return (history, " ➔ ".join([id2name[i] for i in history]), 
                gr.update(choices=recs, value=None, visible=True), 
                b_df, c_df, gr.update(visible=True), gr.update(visible=True), gr.update(visible=False))

    start_btn.click(start_fn, [level, goal, equip], [profile_state, main_app])
    first_ex_drop.change(add_ex_fn, [first_ex_drop, history_state, profile_state], 
                         [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results, add_btn, first_ex_row])
    add_btn.click(add_ex_fn, [recommendations, history_state, profile_state], 
                  [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results, add_btn, first_ex_row])
    reset_btn.click(lambda: ([], "", gr.update(visible=False), gr.update(visible=True)), None, [history_state, history_display, main_app, first_ex_row])

demo.launch(debug=True, share=True)
```

```python
import gradio as gr
import pandas as pd
import torch
import torch.nn as nn
import json
import os
from catboost import CatBoostClassifier
from deep_translator import GoogleTranslator

# --- 1. СЛОВАРИ ПЕРЕВОДА ПРИЗНАКОВ (Маппинг) ---
# Для UI -> RU, для модели -> EN
LEVEL_TRANS = {
    "Новичок": "Novice",
    "Начинающий": "Beginner",
    "Средний": "Intermediate",
    "Продвинутый": "Advanced"
}

GOAL_TRANS = {
    "Бодибилдинг": "Bodybuilding",
    "Рельеф и мышцы": "Muscle&Sculpting",
    "Пауэрлифтинг": "Powerlifting",
    "Тяжелая атлетика": "OlympicWeightlifting",
    "Фитнес (свой вес)": "BodyweightFitness",
    "Общий фитнес": "Fitness",
    "Атлетика": "Athletics",
    "Пауэрбилдинг": "Powerbuilding"
}

EQUIP_TRANS = {
    "Тренажеры": "Machine",
    "Свой вес": "Bodyweight",
    "Гантели": "Dumbbell",
    "Штанга": "Barbell",
    "Фитнес-резинки": "Bands",
    "Кардио": "Cardio"
}

# Обратные словари для отображения в таблицах
REV_EQUIP = {v: k for k, v in EQUIP_TRANS.items()}
REV_GOAL = {v: k for k, v in GOAL_TRANS.items()}

# --- 2. АРХИТЕКТУРА И ЗАГРУЗКА ---
class TransformerRec(nn.Module):
    def __init__(self, vocab_size, hidden=256, heads=8, layers=4, max_len=20, dropout=0.2):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, hidden)
        self.pos_emb = nn.Embedding(max_len, hidden)
        self.dropout = nn.Dropout(dropout)
        encoder_layer = nn.TransformerEncoderLayer(d_model=hidden, nhead=heads, dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=layers)
        self.fc = nn.Linear(hidden, vocab_size)

    def forward(self, x):
        positions = torch.arange(x.size(1), device=x.device).unsqueeze(0)
        x = self.dropout(self.token_emb(x) + self.pos_emb(positions))
        x = self.transformer(x)
        return self.fc(x)

device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
data_folder = "../gym_model"

# Кеш перевода названий упражнений
translator = GoogleTranslator(source='en', target='ru')
translation_cache = {}

def get_ru(text):
    if not text or pd.isna(text): return ""
    if text not in translation_cache:
        try:
            translation_cache[text] = translator.translate(text)
        except: return text
    return translation_cache[text]

with open(f'{data_folder}/vocab.json', 'r') as f:
    vocab = json.load(f)
id2name = {v: k for k, v in vocab.items()}

df_meta = pd.read_csv("../programs_detailed_boostcamp_kaggle.csv")
exercise_meta_dict = df_meta.drop_duplicates(subset=['exercise_name']).set_index('exercise_name').to_dict('index')

model = TransformerRec(vocab_size=len(vocab)).to(device)
bert_path = f'{data_folder}/checkpoints-2/gym_bert_v2_ep27_hit0.3522.pth'
if os.path.exists(bert_path):
    model.load_state_dict(torch.load(bert_path, map_location=device))
    model.eval()

cat_ranker = CatBoostClassifier()
cat_path = f'{data_folder}/cat_final.cbm' 
if os.path.exists(cat_path):
    cat_ranker.load_model(cat_path)

# --- 3. ЛОГИКА ---
def predict_debug(history_ids, user_profile, top_k=10):
    input_seq = torch.LongTensor([history_ids]).to(device)
    with torch.no_grad():
        logits = model(input_seq)
        probs = torch.softmax(logits[0, -1, :], dim=-1)
    
    scores, indices = torch.topk(probs, k=100) 
    candidates = []
    last_ex_name = id2name.get(history_ids[-1], "")
    last_meta = exercise_meta_dict.get(last_ex_name, {})

    # Маппинг уровней для CatBoost
    LEVEL_MAP = {'Novice': 0, 'Beginner': 1, 'Intermediate': 2, 'Advanced': 3}

    for s, idx in zip(scores.cpu().numpy(), indices.cpu().numpy()):
        name = id2name[idx]
        current_meta = exercise_meta_dict.get(name, {})
        is_same = 1 if current_meta.get('category') == last_meta.get('category') else 0
        
        candidates.append({
            'name': name,
            'bert_score': float(s),
            'is_same_group': is_same,
            # Передаем в модель чистые английские значения
            'level_idx': LEVEL_MAP.get(LEVEL_TRANS[user_profile['level']], 1),
            'goal_clean': GOAL_TRANS[user_profile['goal']],
            'eq_clean': EQUIP_TRANS[user_profile['equipment']]
        })
    
    cand_df = pd.DataFrame(candidates)
    features_order = ['bert_score', 'is_same_group', 'level_idx', 'goal_clean', 'eq_clean']
    
    preds = cat_ranker.predict_proba(cand_df[features_order])[:, 1]
    cand_df['final_score'] = preds
    
    final_df = cand_df.sort_values('final_score', ascending=False).head(top_k)
    final_df['name_ru'] = final_df['name'].apply(get_ru)
    
    cat_table = final_df[['name_ru', 'final_score', 'bert_score']].copy()
    cat_table.columns = ['Упражнение', 'Вероятность', 'BERT Скор']
    
    bert_df = pd.DataFrame({
        "Упражнение": [get_ru(id2name[i.item()]) for i in indices[:10]],
        "BERT Скор": [round(s.item(), 4) for s in scores[:10]]
    })
    
    return final_df['name'].tolist(), final_df['name_ru'].tolist(), bert_df, cat_table

# --- 4. ИНТЕРФЕЙС ---
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    history_state = gr.State([])
    profile_state = gr.State({})
    
    gr.Markdown("# 🤖 ИИ-Тренер: Интеллектуальный подбор программы")
    
    with gr.Row(variant='panel'):
        # Используем ключи русских словарей
        level = gr.Dropdown(list(LEVEL_TRANS.keys()), label="Ваш уровень опыта", value="Начинающий")
        goal = gr.Dropdown(list(GOAL_TRANS.keys()), label="Основная цель", value="Бодибилдинг")
        equip = gr.Dropdown(list(EQUIP_TRANS.keys()), label="Доступный инвентарь", value="Гантели")
        start_btn = gr.Button("🚀 Начать тренировку", variant="primary")

    with gr.Column(visible=False) as main_app:
        history_display = gr.Textbox(label="Ваш план (текущая цепочка)", interactive=False)
        
        with gr.Row() as first_ex_row:
            first_ex_drop = gr.Dropdown(choices=sorted(list(vocab.keys())), label="Выберите первое упражнение (на англ.):", interactive=True)
            
        with gr.Row(visible=False) as results:
            with gr.Column():
                gr.Markdown("### 🧠 Поиск по смыслу (BERT)")
                bert_ui_table = gr.DataFrame()
            with gr.Column():
                gr.Markdown("### ⚖️ Фильтр по инвентарю и цели (CatBoost)")
                cat_ui_table = gr.DataFrame()
        
        recommendations = gr.Dropdown(choices=[], label="Выберите следующее упражнение:", interactive=True, visible=False)
        add_btn = gr.Button("➕ Добавить в цепочку", variant="primary", visible=False)
        reset_btn = gr.Button("🔄 Сброс", variant="stop")

    def start_fn(l, g, e):
        return {profile_state: {'level': l, 'goal': g, 'equipment': e}, main_app: gr.update(visible=True)}

    def add_ex_fn(ex_name, history, profile):
        if not ex_name: return history, "", gr.update(), None, None, gr.update(), gr.update()
        
        new_id = vocab[ex_name]
        if not history or history[-1] != new_id: history.append(new_id)
        
        en_names, ru_names, b_df, c_df = predict_debug(history, profile)
        history_ru = [get_ru(id2name[i]) for i in history]
        rec_choices = list(zip(ru_names, en_names))
        
        return (history, " ➔ ".join(history_ru), 
                gr.update(choices=rec_choices, value=None, visible=True), 
                b_df, c_df, gr.update(visible=True), gr.update(visible=True), gr.update(visible=False))

    start_btn.click(start_fn, [level, goal, equip], [profile_state, main_app])
    first_ex_drop.change(add_ex_fn, [first_ex_drop, history_state, profile_state], 
                         [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results, add_btn, first_ex_row])
    add_btn.click(add_ex_fn, [recommendations, history_state, profile_state], 
                  [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results, add_btn, first_ex_row])
    reset_btn.click(lambda: ([], "", gr.update(visible=False), gr.update(visible=True)), None, [history_state, history_display, main_app, first_ex_row])

demo.launch()
```

```python
import gradio as gr
import pandas as pd
import torch
import torch.nn as nn
import json
import os
from catboost import CatBoostClassifier

# --- 1. CONFIG & MAPS ---
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
data_folder = "../gym_model"

# Маппинг уровней (как в обучении)
LEVEL_MAP = {'Novice': 0, 'Beginner': 1, 'Intermediate': 2, 'Advanced': 3}

# --- 2. ARCHITECTURE ---
class TransformerRec(nn.Module):
    def __init__(self, vocab_size, hidden=256, heads=8, layers=4, max_len=20, dropout=0.2):
        super().__init__()
        self.token_emb = nn.Embedding(vocab_size, hidden)
        self.pos_emb = nn.Embedding(max_len, hidden)
        self.dropout = nn.Dropout(dropout)
        encoder_layer = nn.TransformerEncoderLayer(d_model=hidden, nhead=heads, dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=layers)
        self.fc = nn.Linear(hidden, vocab_size)

    def forward(self, x):
        positions = torch.arange(x.size(1), device=x.device).unsqueeze(0)
        x = self.dropout(self.token_emb(x) + self.pos_emb(positions))
        x = self.transformer(x)
        return self.fc(x)

# --- 3. LOADING ---
with open(f'{data_folder}/vocab.json', 'r') as f:
    vocab = json.load(f)
id2name = {v: k for k, v in vocab.items()}

# Метаданные (обязательно должны содержать колонку 'category' для is_same_group)
df_meta = pd.read_csv("../programs_detailed_boostcamp_kaggle.csv")
exercise_meta_dict = df_meta.drop_duplicates(subset=['exercise_name']).set_index('exercise_name').to_dict('index')

# Загрузка BERT
model = TransformerRec(vocab_size=len(vocab)).to(device)
bert_path = f'{data_folder}/checkpoints-2/gym_bert_v2_ep27_hit0.3522.pth'
if os.path.exists(bert_path):
    model.load_state_dict(torch.load(bert_path, map_location=device))
    model.eval()

# Загрузка НОВОЙ модели CatBoost (с локациями)
cat_ranker = CatBoostClassifier()
cat_path = f'{data_folder}/cat_location_v1.cbm' 
if os.path.exists(cat_path):
    cat_ranker.load_model(cat_path)

# --- 4. PREDICTION LOGIC ---
def predict_debug(history_ids, user_profile, top_k=10):
    input_seq = torch.LongTensor([history_ids]).to(device)
    with torch.no_grad():
        logits = model(input_seq)
        probs = torch.softmax(logits[0, -1, :], dim=-1)
    
    # Берем ТОП-300 от BERT (увеличили, чтобы CatBoost было из чего выбирать для Home/Outdoor)
    scores, indices = torch.topk(probs, k=300) 
    candidates = []
    
    last_ex_name = id2name.get(history_ids[-1], "")
    last_meta = exercise_meta_dict.get(last_ex_name, {})

    for s, idx in zip(scores.cpu().numpy(), indices.cpu().numpy()):
        name = id2name[idx]
        current_meta = exercise_meta_dict.get(name, {})
        is_same = 1 if current_meta.get('category') == last_meta.get('category') else 0
        
        candidates.append({
            'name': name,
            'bert_score': float(s),
            'is_same_group': is_same,
            'level_idx': LEVEL_MAP.get(user_profile['level'], 1),
            'goal_clean': user_profile['goal'],
            'location': user_profile['location'] # Передаем НОВУЮ фичу
        })
    
    cand_df = pd.DataFrame(candidates)
    
    # ПОРЯДОК ФИЧ: Должен быть в точности как при model.fit()!
    features_order = ['bert_score', 'is_same_group', 'level_idx', 'goal_clean', 'location']
    
    # Реранжирование
    preds = cat_ranker.predict_proba(cand_df[features_order])[:, 1]
    cand_df['final_score'] = preds
    
    final_df = cand_df.sort_values('final_score', ascending=False).head(top_k)
    
    cat_table = final_df[['name', 'final_score', 'bert_score']].copy()
    cat_table.columns = ['Exercise', 'CatBoost Prob', 'BERT Score']
    
    bert_table = pd.DataFrame({
        "Exercise": [id2name[i.item()] for i in indices[:10]],
        "BERT Score": [round(s.item(), 4) for s in scores[:10]]
    })
    
    return final_df['name'].tolist(), bert_table, cat_table

# --- 5. GRADIO INTERFACE ---
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    history_state = gr.State([])
    profile_state = gr.State({})
    
    gr.Markdown("# 🏋️‍♂️ AI Trainer v2: Context-Aware (Location Based)")
    
    with gr.Row(variant='panel'):
        level = gr.Dropdown(['Novice', 'Beginner', 'Intermediate', 'Advanced'], label="Experience Level", value="Beginner")
        goal = gr.Dropdown(['Bodybuilding', 'Muscle&Sculpting', 'Powerlifting', 'Fitness', 'Athletics'], label="Goal", value="Fitness")
        # ЗАМЕНА: Теперь выбираем ГДЕ тренируемся
        location = gr.Radio(['Home', 'Gym', 'Outdoor'], label="Current Location", value="Home")
        start_btn = gr.Button("🚀 Start Session", variant="primary")

    with gr.Column(visible=False) as main_app:
        history_display = gr.Textbox(label="Your Workout Sequence", interactive=False)
        
        with gr.Row() as first_ex_row:
            first_ex_drop = gr.Dropdown(choices=sorted(list(vocab.keys())), label="Choose Starting Exercise:", interactive=True)
            
        with gr.Row(visible=False) as results_row:
            with gr.Column():
                gr.Markdown("### 🧠 BERT (Semantic Similarity)")
                bert_ui_table = gr.DataFrame()
            with gr.Column():
                gr.Markdown("### ⚖️ CatBoost (Location & Context)")
                cat_ui_table = gr.DataFrame()
        
        recommendations = gr.Dropdown(choices=[], label="AI Recommended Next Steps:", interactive=True, visible=False)
        add_btn = gr.Button("➕ Add to Sequence", variant="primary", visible=False)
        reset_btn = gr.Button("🔄 Reset", variant="stop")

    def start_fn(l, g, loc):
        # Сохраняем локацию в профиль
        return {profile_state: {'level': l, 'goal': g, 'location': loc}, main_app: gr.update(visible=True)}

    def add_ex_fn(ex_name, history, profile):
        if not ex_name: return history, "", gr.update(), None, None, gr.update(), gr.update()
        
        new_id = vocab[ex_name]
        if not history or history[-1] != new_id: history.append(new_id)
        
        recs, b_df, c_df = predict_debug(history, profile)
        history_text = " ➔ ".join([id2name[i] for i in history])
        
        return (history, history_text, 
                gr.update(choices=recs, value=None, visible=True), 
                b_df, c_df, gr.update(visible=True), gr.update(visible=True), gr.update(visible=False))

    start_btn.click(start_fn, [level, goal, location], [profile_state, main_app])
    
    first_ex_drop.change(add_ex_fn, [first_ex_drop, history_state, profile_state], 
                         [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results_row, add_btn, first_ex_row])
    
    add_btn.click(add_ex_fn, [recommendations, history_state, profile_state], 
                  [history_state, history_display, recommendations, bert_ui_table, cat_ui_table, results_row, add_btn, first_ex_row])
    
    reset_btn.click(lambda: ([], "", gr.update(visible=False), gr.update(visible=True)), None, [history_state, history_display, main_app, first_ex_row])

demo.launch(debug=True)
```
